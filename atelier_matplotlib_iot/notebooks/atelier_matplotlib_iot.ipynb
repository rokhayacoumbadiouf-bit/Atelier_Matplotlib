1. Afficher l'évolution de la température en fonction du temps, avec le titre, les labels, la légende 
et la grille. 

```python
plt.figure(figsize=(9, 6))
plt.plot(df['date_heure'], df['temperature'], label='Temperature (°C)', color='orange')
plt.title("Courbe d'evolution de la temperature , en fonction du temps" )
plt.xlabel("Temps", fontsize=10)
plt.ylabel("Temperature",fontsize=10)
plt.grid(True, linestyle=":", alpha=0.6)
plt.legend(loc="upper left", framealpha=0.9)
plt.show()
```

 2.Identifier visuellement les valeurs qui semblent anormales. 

La zone correspondant au 4/6 de la courbe correspond à une zone de températures très élévées , pouvant être considéré anormales.

Partie 2 – Diagramme en barres (Bar Chart) 
1) Déterminer consommation_batiment qui stocke les consommations moyennes des bâtiments 
2) Comparer graphiquement la consommation moyenne des bâtiments 
3) Comparer la consommation moyenne des bâtiments avec un graphique horizontal 
4) Dans quels cas un graphique horizontal peut-il être plus lisible ? 


1.Déterminer consommation_batiment qui stocke les consommations moyennes des bâtiments

```python
consommation_batiment = df.groupby("batiment")["consommation"].mean()
```

2.Comparer graphiquement la consommation moyenne des bâtiments

```python
plt.figure(figsize=(9, 6))
x_abs = consommation_batiment.index
y_ord = consommation_batiment.values
plt.bar(x_abs,y_ord)
plt.title("Consommation moyenne par batiment" )
plt.xlabel("batiment", fontsize=10)
plt.ylabel("consommation moyenne ",fontsize=10)
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

3.Comparer la consommation moyenne des bâtiments avec un graphique horizontal

```python
plt.figure(figsize=(9, 6))
x_abs = consommation_batiment.index
y_ord = consommation_batiment.values
plt.barh(x_abs,y_ord)
plt.title("Consommation moyenne par batiment" )
plt.xlabel("batiment", fontsize=10)
plt.ylabel("consommation moyenne ",fontsize=10)
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

4. Dans quels cas un graphique horizontal peut-il être plus lisible ?

si l’axe des catégories est plus important que l’axe des valeurs, ou si les libellés sont longs, le graphique horizontal est souvent le meilleur choix.

1) Afficher la distribution des températures en divisant la plage des données en 20 classes et en

```python
plt.figure(figsize=(9, 6))
x_abs = df['temperature']
plt.hist(x_abs,bins=20)
plt.title("Distribution des températures sur 20 classes" )
plt.xlabel("Temperatures", fontsize=10)
plt.grid(True,axis="y", linestyle="--", alpha=0.6)
plt.show()
```
