3.Comparer la consommation moyenne des bâtiments avec un graphique horizontal

```python
plt.figure(figsize=(9, 6))
x_abs = consommation_batiment.index
y_ord = consommation_batiment.values
plt.barh(x_abs,y_ord)
plt.title("Consommation moyenne par batiment" )
plt.xlabel("batiment", fontsize=10)
plt.ylabel("consommation moyenne ",fontsize=10)
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```

4. Dans quels cas un graphique horizontal peut-il être plus lisible ?

si l’axe des catégories est plus important que l’axe des valeurs, ou si les libellés sont longs, le graphique horizontal est souvent le meilleur choix.

1)Créer un box plot de la température. N’oublier pas de supprimer les valeurs manquantes avant 
de construire le graphique et afficher uniquement les lignes horizontales de la grille 


```python
temperature = df["temperature"].dropna()
```

```python
plt.figure(figsize=(9,6))

plt.boxplot(temperature)

plt.title("Box plot de la température")
plt.ylabel("Température (°C)")

plt.grid(axis="y")

plt.show()
```

2) Que représentent les points situés en dehors des moustaches ? 

Les points situés en dehors des moustaches représentent généralement des valeurs potentiellement aberrantes selon la règle utilisée par le box plot.

Le box plot utilise notamment :

Q1 ;
médiane ;
Q3 ;
IQR = Q3 − Q1.

Les observations suffisamment éloignées des quartiles sont affichées comme des points individuels.

3)Créer un box plot de la consommation

```python
consommation = df["consommation"].dropna()

plt.figure(figsize=(8, 5))

plt.boxplot(consommation)

plt.title("Box plot de la consommation")
plt.ylabel("Consommation")

plt.grid(axis="y")

plt.show()
```

4)A partir des deux box plot, comparer 

a) la position de la médiane ;  
b) la taille de la boîte ;  
c) la longueur des moustaches ;  
d) les valeurs extrêmes. 

```python
print("TEMPÉRATURE")
print("Médiane :", temperature.median())
print("Q1 :", temperature.quantile(0.25))
print("Q3 :", temperature.quantile(0.75))
print("IQR :", temperature.quantile(0.75) - temperature.quantile(0.25))
print("Min :", temperature.min())
print("Max :", temperature.max())

print("\nCONSOMMATION")
print("Médiane :", consommation.median())
print("Q1 :", consommation.quantile(0.25))
print("Q3 :", consommation.quantile(0.75))
print("IQR :", consommation.quantile(0.75) - consommation.quantile(0.25))
print("Min :", consommation.min())
print("Max :", consommation.max())
```

a) Position de la médiane

La température a une médiane de 24,86 °C. Cela signifie que 50 % des mesures sont inférieures à cette valeur et 50 % sont supérieures.

La consommation a une médiane de 206,15.

b) Taille de la boîte

Pour la température :

IQR = Q3 − Q1 = 27,275 − 22,57 = 4,705 °C

Pour la consommation :

IQR = 254,1275 − 160,1775 = 93,95

La boîte de la consommation est donc beaucoup plus étendue dans son unité de mesure.

c) Longueur des moustaches

Pour la température, l'étendue totale va de -18,5 °C à 58,7 °C, ce qui est très large par rapport à son IQR de seulement 4,705 °C.

Cela suggère la présence de valeurs très éloignées de la zone centrale.

Pour la consommation, l'étendue va de 18,12 à 875, également très importante par rapport à son IQR de 93,95.

d) Valeurs extrêmes

Les valeurs -18,5 °C et 58,7 °C sont particulièrement éloignées de la zone centrale des températures.

De même, la consommation maximale de 875 est très éloignée de la médiane de 206,15.

Cela mérite d'être vérifié dans le dataset : ces valeurs peuvent être de véritables mesures, mais elles peuvent également correspondre à des anomalies de capteur ou à des erreurs de saisie.

5)La consommation présente-t-elle davantage de dispersion que la température ?

La consommation présente un intervalle interquartile de 93,95, tandis que celui de la température est de 4,705 °C. La consommation présente donc une dispersion absolue beaucoup plus importante sur son échelle de mesure. Les deux variables ayant des unités différentes, cette comparaison doit toutefois être interprétée avec prudence.

Partie 6 – Diagramme circulaire (Pie Chart) 
1) Réaliser un diagramme pour analyser la répartition des états (OK, ALERTE, ERREUR) 
2) interpréter la proportion de capteurs OK ; proportion ALERTE ;  proportion ERREUR  


1) Réaliser un diagramme pour analyser la répartition des états (OK, ALERTE, ERREUR) 

```python
repartition_etats = df["etat"].value_counts()

print(repartition_etats)
```

```python
plt.figure(figsize=(10, 8))

plt.pie(
    repartition_etats.values,
    labels=repartition_etats.index,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.75
)

plt.title("Répartition des états des capteurs")

plt.tight_layout()
plt.show()
```

2)interpréter la proportion de capteurs OK ; proportion ALERTE ;  proportion ERREUR 

```python
proportions = df["etat"].value_counts(normalize=True) * 100

print(proportions)
```

une forte proportion OK indique que la majorité des capteurs fonctionne normalement ;
ALERTE indique des capteurs nécessitant une surveillance ;
ERREUR indique des capteurs présentant un problème nécessitant probablement une intervention.

Partie 7 : Plusieurs courbes sur un même graphique 
Construire sur un même graphique  « Évolution de la température par bâtiment" » des quatre 
bâtiments. 

```python
plt.figure(figsize=(13, 6))

for batiment in df["batiment"].dropna().unique():

    donnees_batiment = df[df["batiment"] == batiment]

    plt.plot(
        donnees_batiment["date_heure"].index,
        donnees_batiment["temperature"],
        label=batiment
    )

plt.title("Évolution de la température par bâtiment")
plt.xlabel("Date et heure")
plt.ylabel("Température (°C)")

plt.legend()
plt.grid(True)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Proposer et réaliser toute fonctionnalité pertinente et utile qu’on pourrait rajouter à l’atelier.

Cette visualisation permet de comparer la température moyenne des différents bâtiments et d'établir un classement. Elle permet notamment d'identifier les bâtiments présentant les températures moyennes les plus élevées ou les plus faibles. Cette information peut être utile pour étudier les différences de conditions environnementales entre les bâtiments.

```python
temperature_batiment = df.groupby("batiment")["temperature"].mean()

print(temperature_batiment)
```

```python
temperature_batiment = (
    df.groupby("batiment")["temperature"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))

barres = plt.bar(
    temperature_batiment.index,
    temperature_batiment.values
)

plt.title("Classement des bâtiments selon la température moyenne")
plt.xlabel("Bâtiment")
plt.ylabel("Température moyenne (°C)")

for barre in barres:
    hauteur = barre.get_height()

    plt.text(
        barre.get_x() + barre.get_width() / 2,
        hauteur,
        f"{hauteur:.2f} °C",
        ha="center",
        va="bottom"
    )

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
